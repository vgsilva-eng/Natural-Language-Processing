## News and Earnings Analysis

The NewsAgent and EarningsAgent will provide specialized financial analysis for the selected stock. The EarningsAgent will evaluate company financial data and key performance metrics, while the NewsAgent will retrieve, classify, and summarize recent financial news.

```python
import sys
from pathlib import Path

from dotenv import load_dotenv

# Add the project root to the Python path
repo_root = Path.cwd().parent
if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

# Load API keys from the local .env file
load_dotenv(repo_root / ".env")

from agents.earnings_agent import EarningsAgent
from agents.news_agent import NewsAgent

print("Setup complete.")
```

### EarningsAgent

The EarningsAgent will retrieve company financial data and evaluate key metrics such as revenue, net income, earnings per share, valuation ratios, and net profit margin. The results will provide a structured view of the company's financial performance.

```python
earnings_agent = EarningsAgent()

earnings_result = earnings_agent.analyze("AAPL")

print(f"Ticker: {earnings_result['ticker']}")
print(f"Sector: {earnings_result['sector']}")
print(f"Industry: {earnings_result['industry']}")
print(f"Revenue: ${earnings_result['revenue']:,.0f}")
print(f"Net Income: ${earnings_result['net_income']:,.0f}")
print(f"EPS: ${earnings_result['eps']:.2f}")
print(f"P/E Ratio: {earnings_result['pe_ratio']:.2f}")
print(f"Forward P/E: {earnings_result['forward_pe']:.2f}")
print(f"Net Profit Margin: {earnings_result['net_margin_pct']:.2f}%")

print("\nFinancial Signals:")
for signal in earnings_result["signals"]:
    print(f"- {signal}")
```

### NewsAgent

The NewsAgent will retrieve recent financial news for the selected stock, classify each article by financial topic, and summarize the key themes. The analysis will highlight relevant developments, market sentiment, risks, and opportunities.

```python
news_agent = NewsAgent()

news_articles = news_agent.fetch_and_classify(
    "AAPL",
    lookback_days=7
)

print(f"Articles Classified: {len(news_articles)}\n")

for article in news_articles[:5]:
    print(f"Title: {article['title']}")
    print(f"Category: {article['category']}")
    print(f"Source: {article['source']}")
    print()
```

```python
news_summary = news_agent.summarize(news_articles)

print("News Summary:\n")
print(news_summary)
```

### Combined Specialist Analysis

The EarningsAgent and NewsAgent will provide complementary information for the investment research process.
