# MarketAgent demo (Obinna)
This notebook demonstrates the market specialist and price chart for the team project. Run it from the repository root after installing `requirements.txt`. It fetches live Yahoo Finance data; results will change by date. Historical signals are descriptive, not forecasts.

```python
from pathlib import Path
import sys
repo_root = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(repo_root))
from IPython.display import SVG, display
from agents.market_agent import MarketAgent
from tools.market_visualization import save_price_chart

ticker = "AAPL"  # Change this to another listed stock symbol.
market = MarketAgent().analyze(ticker, period="3mo")
print(f"{market['ticker']} as of {market['as_of']}")
print(market['summary'])
print(f"Current price: ${market['current_price']:,.2f}")
print(f"50-day average: {market['ma_50']}")
print(f"200-day average: {market['ma_200']}")
```

```python
chart_path = save_price_chart(market, repo_root / "market_price_chart.svg")
display(SVG(filename=str(chart_path)))
```

## How this fits the team workflow
The routing workflow calls `MarketAgent.analyze(ticker)` for price and technical questions. The result includes `summary`, individual signals, source, observation date, and underlying history. The ResearchAgent can use the summary during synthesis, while the chart can be included in the final HTML or PDF notebook. A 3-month window usually cannot produce a 200-session moving average; `None` is the expected result.
